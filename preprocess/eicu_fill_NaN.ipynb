## TRAIN-eicu_heartrate_systemicmean

```python
import torch

# Load the TEST eICU dataset
data_path = '../data/eicu/TRAIN-eicu_heartrate_systemicmean.pt'
data = torch.load(data_path)

# Display basic information about the loaded data
print(f"Data type: {type(data)}")
print(f"Data shape: {data.shape if hasattr(data, 'shape') else 'N/A'}")
print(f"Data: {data}")
```

15717 patients, 9 channels, 276 timestamps

```python
# Assuming "columns" refer to the channels (axis 1 in the 3D tensor)
nan_per_channel = torch.isnan(data).sum(dim=[0, 2])
print(nan_per_channel)
```

```python
# Impute NaN values using forward fill along the time axis (timestamps)
for patient in range(data.shape[0]):
  for channel in range(data.shape[1]):
    ts = data[patient, channel, :]
    # If the first value is NaN, fill with the mean of the channel (across all patients and times, ignoring NaNs)
    if torch.isnan(ts[0]):
      channel_mean = torch.nanmean(data[:, channel, :])
      ts[0] = channel_mean
    for t in range(1, ts.shape[0]):
      if torch.isnan(ts[t]):
        ts[t] = ts[t-1]
```

```python
# Assuming "columns" refer to the channels (axis 1 in the 3D tensor)
nan_per_channel = torch.isnan(data).sum(dim=[0, 2])
print(nan_per_channel)
```

```python
torch.save(data, data_path)
print(f"Resaved data to {data_path}")
```
